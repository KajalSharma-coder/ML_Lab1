```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,4))
sns.boxplot(y=df['CGPA'])

plt.title("Box Plot of CGPA")
plt.ylabel("CGPA")
plt.show()
```

```python
# Calculate Q1, Q3 and IQR
Q1 = df['CGPA'].quantile(0.25)
Q3 = df['CGPA'].quantile(0.75)
IQR = Q3 - Q1

# Calculate lower and upper limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
outliers = df[(df['CGPA'] < lower_limit) | (df['CGPA'] > upper_limit)]

print("Number of Outliers:", len(outliers))
outliers
```

```python
outliers[['StudentID', 'CGPA']]
```

```python
df_no_outliers = df[(df['CGPA'] >= lower_limit) & (df['CGPA'] <= upper_limit)]

print("Original Shape:", df.shape)
print("After Removing Outliers:", df_no_outliers.shape)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,4))
sns.boxplot(y=df_no_outliers['CGPA'])

plt.title("Box Plot of CGPA")
plt.ylabel("CGPA")

plt.show()
```
